```python
# Configuración del entorno
import os
import sys
sys.path.append(os.path.abspath('..'))  # Para importar desde src
import pandas as pd
```

```python
# Carga de datos con pandas desde src
from src.carga_numpy import cargar_datos
ruta_archivo = '../data/retail_sales_dataset.csv'
df = pd.read_csv(ruta_archivo)
df.head()
```

```python
#Total de ventas por categoria
ventas_por_categoria = df.groupby("Product Category")["Total Amount"].sum().sort_values(ascending=False)
ventas_por_categoria
```

```python
#promedio de ventas diarias por categoria
df["Date"] = pd.to_datetime(df["Date"])
ventas_diarias = df.groupby(["Date", "Product Category"])["Total Amount"].sum().reset_index()
promedio_ventas_diarias = ventas_diarias.groupby("Product Category")["Total Amount"].mean().sort_values(ascending=False)
promedio_ventas_diarias
```

```python
#categorias con la mayora y menor cantidad de ventas
categoria_mayor_venta = ventas_por_categoria.idxmax()
categoria_menor_venta = ventas_por_categoria.idxmin()
print("Categoría con mayor venta:", categoria_mayor_venta)
print("Categoría con menor venta:", categoria_menor_venta)
```

```python
#filtro por categoria especifica
filtro_electronics = df[df['Product Category'] == 'Electronics']
filtro_electronics.head()
```

```python
#Estadisticas adicionales
# Suma total de ventas
total_ventas = df['Total Amount'].sum()
# Precio promedio por unidad
precio_promedio = df['Price per Unit'].mean()
# Cantidad total de productos vendidos
cantidad_total = df['Quantity'].sum()
total_ventas, precio_promedio, cantidad_total
```
